```python
fig, ax = plt.subplots()
im = ax.matshow(score_dtw_genormaliseerd, cmap=plt.cm.Blues)
ax.set_xticks(np.arange(len(time_deltas)), labels=time_deltas)
ax.set_yticks(np.arange(len(radiuses)), labels=radiuses)
ax.set_xlabel("Time Delta")
ax.set_ylabel("Radius")
ax.set_title("Score Evaluatie functie", pad=20)  # Hoofdtitel
cbar = plt.colorbar(im, ax=ax)
cbar.set_label("Score")
for i in range(len(radiuses)):
    for j in range(len(time_deltas)):
        # Bepaal of de achtergrond licht of donker is (kies zwart/wit tekst)
        cell_value = round(score_dtw_genormaliseerd[i][j])
        rgba = im.cmap(im.norm(cell_value))  # RGB-kleur van de cel
        brightness = (
            0.299 * rgba[0] + 0.587 * rgba[1] + 0.114 * rgba[2]
        )  # Helderheid berekenen
        text_color = (
            "black" if brightness > 0.6 else "white"
        )  # Drempelwaarde aanpassen indien nodig

        ax.text(
            j,
            i,
            f"{cell_value}",
            ha="center",
            va="center",
            color=text_color,
            fontsize=8,
        )
```
